```python
# ============================================================
# STEP 2: Load Insecticide Resistance Data (IR Mapper)
# ============================================================
import pandas as pd
import requests
import os
import warnings
warnings.filterwarnings('ignore')

# Create data folder
os.makedirs("C:/malaria_resistance/data", exist_ok=True)

# Download IR Mapper dataset (WHO bioassay results - public)
url = "https://raw.githubusercontent.com/malaria-atlas-project/ir-mapper/master/data/ir_mapper_data.csv"

print("Downloading IR Mapper resistance data...")
response = requests.get(url, timeout=30)

if response.status_code == 200:
    with open("C:/malaria_resistance/data/ir_mapper.csv", "wb") as f:
        f.write(response.content)
    print("Downloaded successfully ✓")
    df = pd.read_csv("C:/malaria_resistance/data/ir_mapper.csv")
    print(f"Rows: {len(df):,}")
    print(f"Columns: {list(df.columns)}")
else:
    print(f"Failed: {response.status_code} - trying backup source...")
    
    # Backup: WHO insecticide resistance data via VectorBase
    url2 = "https://vectorbase.org/common/downloads/Current_Release/insecticide_resistance.csv"
    response2 = requests.get(url2, timeout=30)
    print(f"Backup status: {response2.status_code}")
```

```python
# ============================================================
# STEP 2: Download Resistance Data - WHO & VectorBase Sources
# ============================================================
import requests
import os

os.makedirs("C:/malaria_resistance/data", exist_ok=True)

# Try VectorBase insecticide resistance download
sources = [
    "https://vectorbase.org/common/downloads/insecticide_resistance/insecticide_resistance_data.csv",
    "https://data.vectorbase.org/downloads/insecticide_resistance.csv",
    "https://www.who.int/docs/default-source/malaria/insecticide-resistance/insecticide-resistance-data.csv",
]

for url in sources:
    print(f"Trying: {url}")
    try:
        r = requests.get(url, timeout=15)
        print(f"  Status: {r.status_code}")
        if r.status_code == 200:
            print("  SUCCESS!")
            break
    except Exception as e:
        print(f"  Error: {e}")
```
